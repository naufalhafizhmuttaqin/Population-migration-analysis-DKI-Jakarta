**Import Library**

```python
import pandas as pd
import numpy as np
import os

# Display all columns
pd.set_option('display.max_columns', None)
```

**Load Dataset**

```python
tables = pd.read_html(
    "/content/Data_Penduduk Pendatang ke Provinsi DKI Jakarta - tabel.xls"
)

print(f"Jumlah tabel: {len(tables)}")
df = tables[0]
```

**Check Missing Values**

```python
df.isnull().sum()
```

```python
missing = (
    df.isnull().sum()
      .to_frame("Missing")
)

missing["Percentage"] = (
    missing["Missing"] / len(df) * 100
)

missing.sort_values("Percentage", ascending=False)
```

**Check Duplicate Rows**

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

**Check Unique Values**

```python
for col in df.columns:
    print("="*50)
    print(col)
    print(df[col].nunique())
```

**Statistical Summary**

```python
df.describe()
```

```python
df.describe(include="all")
```

**Check Invalid Values**

```python
df[df["jumlah"] < 0]
```

**Check Whitespace**

```python
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    df[col] = df[col].str.strip()
```

**Standardize Text**

```python
for col in text_cols:
    df[col] = df[col].str.title()
```

**Rename Columns**

```python
df.columns = [
    "year",
    "destination_city",
    "destination_district",
    "destination_village",
    "origin_province",
    "origin_city",
    "total_migrants"
]
```
